### Данные
Однажды был проведен эксперимент - на 5000 участниках проверяли, как возраст и физическое здоровье влияют на потерю слуха, особенно на способность слышать высокие тона. 
 - Признаки
   - age - возраст участников в годах
   - physical_score - оценка физическаого здоровья, полученная в результате физических тестов
 - Целевая переменная (Label/Target)
   - test_results - 0 если аудио-тест не пройден и 1 - если аудио-тест пройден
