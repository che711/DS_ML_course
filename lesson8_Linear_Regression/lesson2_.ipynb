```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../course_data/08-Linear-Regression-Models/Advertising.csv')
df.head()
```

```python
# разбиваем данные на признаки и целевую переменную
X = df.drop('sales',axis=1)
y = df['sales']
```

```python
from sklearn.preprocessing import PolynomialFeatures
```

```python
polynomial_converter = PolynomialFeatures(degree=3, include_bias=False)
```

```python
poly_features = polynomial_converter.fit_transform(X)
```

```python
poly_features.shape
```

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(poly_features, y, test_size=0.3, random_state=101)
```

```python
X_train.shape
```

```python
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train = scaler.transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
X_train[20]
```

```python
poly_features[0]
```

### Ridge-регрессия (l2 regularization) - это метод регуляризацииб позволяющий снизить вероятность  переобученности модели (overfitting) на обучающем наборе данных (после кроссвалидации). Это достигается добавлемтем штрафующего слагаемого к ошибке, с квадратом значений коэффициентов



```python
from sklearn.linear_model import Ridge
```

```python
# help(Ridge)
```

```python
ridge_model = Ridge(alpha=10)
```

```python
ridge_model.fit(X_train, y_train)
```

```python
test_prediction = ridge_model.predict(X_test)
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error
```

```python
MAE = mean_absolute_error(y_test, test_prediction)
MAE
```

```python
RMSE = np.sqrt(mean_squared_error(y_test, test_prediction))
RMSE
```

```python
from sklearn.linear_model import RidgeCV
```

```python
ridge_cv_model = RidgeCV(alphas=(0.1,1.0,10.0), scoring='neg_mean_absolute_percentage_error')
```

```python
ridge_cv_model.fit(X_train, y_train)
```

```python
ridge_cv_model.alpha_
```

```python
from sklearn.metrics._scorer import _SCORERS
```

```python
_SCORERS.keys()
```

```python
test_prediction = ridge_cv_model.predict(X_test)
```

```python
MAE = mean_absolute_error(y_test, test_prediction)
MAE
```

```python
RMSE = np.sqrt(mean_squared_error(y_test, test_prediction))
RMSE
```

```python
ridge_cv_model.coef_
```

```python
ridge_cv_model.best_score_ 
```
